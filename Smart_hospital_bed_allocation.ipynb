```python
#data entry
import pandas as pd
import matplotlib.pyplot as plt
import statistics
pateint_name=["hari","gopi","lakshimi","priya","praveen","joythi","pravalika"]
age=[26,40,20,21,68,65,80]
condition=["normal","critical","normal","good","critical","good","critical"]
oxygen_lev=[91,80,90,89,85,88,80]
avb_beds=4
df=pd.DataFrame({
 "patiant name":pateint_name,
 "age:":age,
 "condition:":condition,
 "oxygen level:":oxygen_lev,
 "available beds:":avb_beds
})
print(df)
```

```python
priority=[]
for index, row in df.iterrows():
    current_condition = row['condition:'].lower()
    current_age = row['age:']
    current_oxygen_lev = row['oxygen level:']
    patient_name = row['patiant name']
    if current_condition == "critical" and current_age > 40 and current_oxygen_lev < 87:
        global avb_beds
        if avb_beds > 0:
            priority.append("High Priority (Bed Allocated)")
            print(f"{patient_name} -> High Priority: Bed Allocated. Available beds before: {avb_beds}")
            avb_beds -= 1
            print(f"Available beds after: {avb_beds}")
        else:
            priority.append("High Priority (No beds available)")
            print(f"{patient_name} -> High Priority: No beds available. Add to waiting list.")
    elif current_condition == "critical":
        priority.append("High Priority (Monitor closely)")
        print(f"{patient_name} -> High Priority: Monitor closely")
    elif current_condition == "good":
        priority.append("Medium Priority (Monitor patient)")
        print(f"{patient_name} -> Medium Priority: Monitor patient")
    elif current_condition == "normal" and (current_age > 60 or current_oxygen_lev < 90):
        priority.append("Medium Priority (Further assessment)")
        print(f"{patient_name} -> Medium Priority: Further assessment")
    else:
        priority.append("Low Priority (Home care)")
        print(f"{patient_name} -> Low Priority: Home care")
```

```python
# Display the final patient priorities and bed allocation status
print("--- Patient Prescriptions ---")
for i, patient_name_from_df in enumerate(df['patiant name']):
    print(f"{patient_name_from_df}: {priority[i]}")
print(f"\nRemaining available beds: {avb_beds}")
```
