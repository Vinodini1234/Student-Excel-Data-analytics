```python
#data set for smart water tank

import matplotlib.pyplot as plt
import pandas as pd
days=[1,2,3,4,5,6,7]
consumption=[25,35,45,55,65,75,85]
data=pd.DataFrame({
    "Day": days,
    "Consumption(L)": consumption,
})
print("Smart Water Tank Dataset (Simplified):")
print(data.head())
```

```python
#water consumption analysis

avg_consumption=sum(consumption)/len(consumption)
print("average consumption of water:",avg_consumption)
```

```python
total_con=sum(consumption)
print("total water used:",total_con)
```

```python
#sum of water incresed daily
daily_increased_water=sum(consumption[1:])-sum(consumption[:-1])
print("water level increased:",daily_increased_water)
```

```python
per_day=data["Daily Increase (L)"] = data["Consumption(L)"].diff().fillna(0)

print("\nWater Level Increase Per Day:",per_day)
```

```python
# sum of water usage increased
water_usage_increased=consumption[2]-consumption[1]
print("water usage increased :",water_usage_increased)
```

```python
#prediction
predict=daily_increased_water+consumption[-1]
print("prediction:",predict)
if daily_increased_water<avg_consumption:
  print("no issue occur")
else:
  print("water sortage occurs")
```

```python
plt.bar(data["Day"], data["Daily Increase (L)"], color="skyblue")
plt.title("Daily Water Level Increase")
plt.xlabel("Day")
plt.ylabel("Increase (Liters)")
plt.show()
```

```python
plt.plot(data["Day"], data["Consumption(L)"], marker='o')
plt.title("Daily Water Consumption")
plt.xlabel("Day")
plt.ylabel("Consumption (Liters)")
plt.grid(True)
plt.show()
```
