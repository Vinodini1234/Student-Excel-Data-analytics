```
#Data intering
import statistics
import matplotlib.pyplot as plt
customers=["hari","gopi","latha","lucky","priya","mdhavi"]
product=["Chair","bad","sopha set","dinning set","study set","tv stand"]
no_of_item=[4,1,3,1,2,1]
product_cost=[400,10000,7000,5000,1500,900]
Delivery_charges=[40,500,400,500,50,100]
internet_in_gb=[4,1,2,4,1,5]
payment_failed=[1,4,2,0,3,0]
purchase_completed=[3,1,1,1,0,0]
print("coustomers:",customers)
print("products:",product)
print("total items:",no_of_item)
print("cost of the product:",product_cost)
print("extra charges:",Delivery_charges)
print("internet speed:",internet_in_gb)
print("total payments failed:",payment_failed)
print("number of otems sold:",purchase_completed)
for i in range(len(customers)):
  print(customers[i],
        no_of_item[i],
        product_cost[i],
        Delivery_charges[i],
        internet_in_gb[i],
        payment_failed[i],
        purchase_completed[i])

```

```
#total abandment
pending=purchase_completed.count(0)
print("total abondment:",pending)
```

```
#internet diagnostic
print(" INTERNET DIAGNOSTIC REPORT ")
total_internet = sum(internet_in_gb)
average_internet = sum(internet_in_gb) / len(internet_in_gb)
max_internet = max(internet_in_gb)
min_internet = min(internet_in_gb)
highest_user = customers[internet_in_gb.index(max_internet)]
lowest_user = customers[internet_in_gb.index(min_internet)]
print("Total Internet Usage (GB):", total_internet)
print("Average Internet Usage (GB):", average_internet)
print("Highest Internet Usage (GB):", max_internet)
print("Lowest Internet Usage (GB):", min_internet)
print("Customer with Highest Internet Usage:", highest_user)
print("Customer with Lowest Internet Usage:", lowest_user)

```

```
#payment diagnostic
total_failed = sum(payment_failed)
total_completed = sum(purchase_completed)
print("Total Payment Failures:", total_failed)
print("Total Purchases Completed:", total_completed)
```

```
#delivery diagnostics
average_delivery = sum(Delivery_charges) / len(Delivery_charges)
print("Average Delivery:", average_delivery)
for charge in Delivery_charges:
    if charge > 50:
        Delivery = "high"
    else:
        Delivery = "fair"
    print(f"Charge: {charge}, Status: {Delivery}")
```

```
#root cause
if total_failed > 5:
    root_cause = "Payment Gateway Issues"
elif average_internet < 5:
    root_cause = "Slow Internet"
elif average_delivery > 50:
    root_cause = "High Delivery Charges"
else:
    root_cause = "No Major Issues"
print("Root Cause:", root_cause)
```

```
#grapical representation
import matplotlib.pyplot as plt
# Customers vs Product Cost
plt.figure(figsize=(10,6))
plt.bar(customers, product_cost, color='skyblue')
plt.title("Product Cost per Customer")
plt.xlabel("Customers")
plt.ylabel("Product Cost")
plt.show()

# Customers vs Delivery Charges
plt.figure(figsize=(10,6))
plt.bar(customers, Delivery_charges, color='orange')
plt.title("Delivery Charges per Customer")
plt.xlabel("Customers")
plt.ylabel("Delivery Charges")
plt.show()

# Payment Failures vs Purchases Completed
plt.figure(figsize=(10,6))
plt.plot(customers, payment_failed, marker='o', label="Payment Failed", color='red')
plt.plot(customers, purchase_completed, marker='o', label="Purchases Completed", color='green')
plt.title("Payment Diagnostics")
plt.xlabel("Customers")
plt.ylabel("Count")
plt.legend()
plt.show()

# Internet Usage
plt.figure(figsize=(10,6))
plt.bar(customers, internet_in_gb, color='purple')
plt.title("Internet Usage per Customer")
plt.xlabel("Customers")
plt.ylabel("Internet (GB)")
plt.show()
```
